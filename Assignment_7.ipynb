```python
import numpy as np
x = np.array([12.0, 18.0, 9.0, 21.0, 15.0])
norm = (x - x.mean()) / x.std()
print(round(norm.mean(), 6))
print(round(norm.std(), 6))

m= np.array([[1, 2, 3], [4, 5, 6]])

print(m.shape)
print(m.mean(axis=0))
print(m.mean(axis=1))
```

```python
import os
import pandas as pd
os.makedirs("data", exist_ok=True)
coffee = [
  {"day": "Mon", "drink": "espresso", "cups": 12, "revenue": 36.0},
  {"day": "Mon", "drink": "latte", "cups": 18, "revenue": 90.0},
  {"day": "Tue", "drink": "espresso", "cups": 10, "revenue": 30.0},
  {"day": "Tue", "drink": "latte", "cups": 22, "revenue": 110.0},
  {"day": "Wed", "drink": "espresso", "cups": 14, "revenue": 42.0},
  {"day": "Wed", "drink": "tea", "cups": 9, "revenue": 27.0},
  {"day": "Thu", "drink": "latte", "cups": 20, "revenue": 100.0},
  {"day": "Fri", "drink": "espresso", "cups": 16, "revenue": 48.0},
  {"day": "Fri", "drink": "latte", "cups": 25, "revenue": 125.0},
  {"day": "Sat", "drink": "latte", "cups": 30, "revenue": 150.0},
  {"day": "Sat", "drink": "tea", "cups": 12, "revenue": 36.0},
  ]
pd.DataFrame(coffee).to_csv("data/coffee\_demo.csv", index=False)
df = pd.read_csv("data/coffee\_demo.csv")
print(df.shape)
print(df["drink"].nunique())
print(round(df["cups"].mean(), 1))
print(df["revenue"].sum())
best_day = df.groupby("day")["revenue"].sum().idxmax()
print(best_day)
```

```python
fri = df[df["day"] == "Fri"].sort_values("revenue", ascending=False)
print(fri["drink"].iloc[0])
```

```python
by_day = df.groupby("day")["revenue"].sum().sort_values(ascending=False)
by_drink = df.groupby("drink")["cups"].mean()
print(by_day.idxmax(), by_day.max())
print(round(by_drink["latte"], 1))
```
