```python
import random
target = random.randint(1, 50)
tries = 0
guess = int(input("خمن رقم من 1 لـ 50: "))
tries += 1
while guess != target:
  if guess > target:
    print("أكتر")
else:
    print("أقل")
guess = int(input("خمن رقم من 1 لـ 50: "))
tries += 1
print(f"برافو! عدد المحاولات: {tries}")
```

```python
scores =[82,90,76,88,95]
total = 0
count = 0
biggest = scores[0]
for score in scores:
  total += score
  count += 1
if score > biggest:
  biggest = score

average = total / count
print(total, round(average, 2), biggest)
```
